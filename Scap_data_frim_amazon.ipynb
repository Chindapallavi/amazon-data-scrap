# Import Required Lib

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import re
import time
```

## GET URL FROM AMAZON


```python
url ="https://www.amazon.in/s?k=laptop&crid=24DV2MXXCSJPH&sprefix=laptop%2Caps%2C394&ref=nb_sb_noss_2"
```

```python
# MAKE HEADER TO MIMIC A BROWSER VISIT
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36 Edg/145.0.0.0"
}
```

```python
data=[]
```

```python
response=requests.get(url,headers=headers)
response.status_code
```

```python
for page in range(1,2):
    params = {'k': 'laptop', 'page': page}
    # make the request with parameters
    response = requests.get(url, headers=headers, params=params)
 
    # beautifulsoup object
    soup = BeautifulSoup(response.content, 'html.parser')
 
    # find all product containers
    products = soup.find_all("div", {"data-component-type": "s-search-result"})
 
    # extract all product details
    for product in products:
        # Extract the product title
        title_tag = product.find("h2")
        if not title_tag:
            continue
        title_text = title_tag.get_text(strip=True)
        # Extract the product price
        price_tag = product.find("span", {"class": "a-price-whole"})
        price_text = price_tag.get_text(strip=True) if price_tag else "N/A"
        #extract Brand
        match=re.match(r'^\W*([A-Za-z]+)', title_text)
        brand=match.group(1).upper() if match else "N/A"
        # extract ram
        ram_match = re.search(r'(\d+)\s*GB\s*(RAM|DOR\d+|LPDDR\d+)', title_text, re.IGNORECASE)
        ram_text = ram_match.group(1) +"GB" if ram_match else "N/A"
        # extract rating
        rating_tag = product.find("span", class_="a-icon-alt")
        rating = "N/A"
        if rating_tag:
            rating_match = re.search(r'(\d+\.?\d*)', rating_tag.get_text())
            rating = rating_match.group(1) if rating_match else "N/A"
        #extract storage
        storage_match = re.search(r'(\d+)\s*(GB|TB)\s*(SSD|HDD)', title_text, re.IGNORECASE)
        storage_text = f"{storage_match.group(1)}{storage_match.group(2)} {storage_match.group(3)}" if storage_match else "N/A"
        #


 
        data.append({
            "Title": title_text,
            "Price": price_text,
            "Brand": brand,
            "RAM": ram_text,
            "Rating": rating,
            "Storage": storage_text
           
        })
 
     
    print(f"Page {page} scraped")
    time.sleep(1)
```

```python
print(len(data))
# print the data
for product in data:
    print("Title:", product.get("Title", "N/A"))
    print("Price:", product.get("Price", "N/A"))
    print("Brand:", product.get("Brand", "N/A"))
    print("RAM:", product.get("RAM", "N/A"))
    print("Rating:", product.get("Rating", "N/A"))
    print("Storage:", product.get("Storage", "N/A"))
    print('-' * 50)
```
